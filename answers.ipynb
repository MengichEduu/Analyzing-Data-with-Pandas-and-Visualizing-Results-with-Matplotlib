```python
# I will Load the dataset called iris.csv
df = pd.read_csv('iris.csv')
```

```python
# Display the first few rows of the dataset using .head() to inspect the data.
print(df.head())
```

```python
# Explore the structure of the dataset by checking the data types and any missing values.
print(df.info())
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Then I will clean the dataset
df_cleaned = df.dropna()
```

```python
# Fill all numeric missing values with their mean
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].mean())
```

```python
# Tak NUMBER 2 STARTS HERE
# Compute descriptive statistics for numerical columns
stats = df.describe()
print("Basic statistics of numerical columns:")
print(stats)
```

```python
# Perform groupings on a categorical column (for example, species, region, or department) and compute the mean of a numerical column for each group
grouped_means = df.groupby('species')['sepal_length'].mean()
print("\nMean sepal length for each species:")
print(grouped_means)
```

```python
print(df.columns)
```

```python
# Group by 'variety' and compute mean of 'sepal.length'
grouped_means = df.groupby('variety')['sepal.length'].mean()
print("\nMean sepal length for each variety:")
print(grouped_means)
```

```python
# TASK NUMBER 3 BEGINS HERE
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
# Calculate the mean sepal length for each variety
mean_sepal_length = df.groupby('variety')['sepal.length'].mean().reset_index()

# Plotting
plt.figure(figsize=(8, 5))
sns.lineplot(data=mean_sepal_length, x='variety', y='sepal.length', marker='o')
plt.title('Average Sepal Length per Variety')
plt.xlabel('Variety')
plt.ylabel('Average Sepal Length')
plt.legend(['Sepal Length'])
plt.show()
```

```python
# Bar chart showing the comparison of a numerical value across categories (e.g., average petal length per species).
avg_petal_length = df.groupby('variety')['petal.length'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=avg_petal_length, x='variety', y='petal.length', palette='viridis')
plt.title('Average Petal Length per Variety')
plt.xlabel('Variety')
plt.ylabel('Average Petal Length')
plt.show()
```

```python
#Histogram of a numerical column to understand its distribution.
plt.figure(figsize=(8, 5))
sns.histplot(df['sepal.length'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
#Scatter plot to visualize the relationship between two numerical columns (e.g., sepal length vs. petal length).
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='sepal.length', y='petal.length', hue='variety', style='variety', palette='deep')
plt.title('Sepal Length vs. Petal Length by Variety')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.legend(title='Variety')
plt.show()
```
